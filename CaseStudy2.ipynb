```python
patient_name = input("Enter patient name: ")

requested = input("Enter requested departments: ").split(",")
available = input("Enter available departments: ").split(",")
previous = input("Enter previously visited departments: ").split(",")
preferred_doctors = input("Enter preferred doctors: ").split(",")
available_doctors = input("Enter available doctors: ").split(",")
emergency = input("Enter emergency departments: ").split(",")

requested = [x.strip() for x in requested]
available = [x.strip() for x in available]
previous = [x.strip() for x in previous]
preferred_doctors = [x.strip() for x in preferred_doctors]
available_doctors = [x.strip() for x in available_doctors]
emergency = [x.strip() for x in emergency]

print("\nFirst requested department:", requested[0])
print("First two departments:", requested[:2])

requested.append("General")
print("After adding:", requested)

requested.remove("General")
print("After removing:", requested)

requested_set = set(requested)
available_set = set(available)
previous_set = set(previous)
emergency_set = set(emergency)

common = requested_set.intersection(available_set)
unavailable = requested_set.difference(available_set)
all_departments = requested_set.union(available_set)
visited = requested_set.intersection(previous_set)
emergency_dept = requested_set.intersection(emergency_set)

duplicates = []

for x in requested:
    if requested.count(x) > 1 and x not in duplicates:
        duplicates.append(x)

preferred_available = []

for doctor in preferred_doctors:
    if doctor in available_doctors:
        preferred_available.append(doctor)

if len(emergency_dept) > 0:
    recommended = list(emergency_dept)[0]
    status = "Emergency Appointment Confirmed"
elif len(common) > 0:
    recommended = list(common)[0]
    status = "Appointment Confirmed"
else:
    recommended = "Not Available"
    status = "No Department Available"

print("\n========== APPOINTMENT REPORT ==========")
print("Patient Name:", patient_name)
print("Requested Departments:", requested)
print("Available Departments:", available)
print("Common Departments:", list(common) if common else "Not Available")
print("Unavailable Departments:", list(unavailable) if unavailable else "Not Available")
print("Previous Departments:", list(visited) if visited else "Not Available")
print("Duplicate Requests:", duplicates if duplicates else "Not Available")
print("Emergency Departments:", list(emergency_dept) if emergency_dept else "Not Available")
print("Preferred Doctors Available:", preferred_available if preferred_available else "Not Available")
print("Recommended Department:", recommended)
print("Final Appointment Status:", status)
print("All Departments:", list(all_departments))
```
