```python
def main():
    print("--- Swiggy / Zomato Order Decision System ---")

    order_amount = float(input("Enter order amount (Rs): "))
    delivery_distance = float(input("Enter distance in km: "))
    customer_type = input("Enter customer type (VIP/Regular/New): ")
    customer_rating = float(input("Enter customer rating (1-5): "))
    restaurant_rating = float(input("Enter restaurant rating (1-5): "))
    prep_time = int(input("Enter prep time in mins: "))
    payment_method = input("Enter payment method (Online/COD): ")
    weather_condition = input("Enter weather (Clear/Rainy/Stormy): ")
    demand_level = input("Enter demand level (Low/Medium/High): ")
    peak_choice = input("Is it peak hour? (yes/no): ")
    previous_cancellations = int(input("Enter past cancellations count: "))

    if peak_choice.lower() == "yes" or peak_choice.lower() == "y":
        is_peak_hour = True
    else:
        is_peak_hour = False

    if restaurant_rating < 3.0 or prep_time > 50:
        restaurant_status = "Overloaded"
    elif prep_time > 35:
        restaurant_status = "Busy"
    else:
        restaurant_status = "Normal"

    if weather_condition.lower() == "stormy" or restaurant_rating < 2.5 or previous_cancellations > 5:
        order_status = "Rejected"
        manual_review = "Not Needed"
    elif order_amount > 4000 or previous_cancellations >= 3 or (delivery_distance > 15 and weather_condition.lower() == "rainy"):
        order_status = "Manual Review"
        manual_review = "Send to Agent"
    else:
        order_status = "Accepted"
        manual_review = "Not Needed"

    if delivery_distance <= 5:
        del_charge = 30.0
    else:
        del_charge = 30.0 + (delivery_distance - 5) * 6.0

    surge = 0.0
    if weather_condition.lower() == "rainy":
        surge = surge + 25.0
    elif weather_condition.lower() == "stormy":
        surge = surge + 50.0

    if is_peak_hour == True:
        surge = surge + 15.0

    if demand_level.lower() == "high":
        surge = surge + 15.0

    total_delivery_fee = del_charge + surge

    discount = 0.0
    if customer_type.lower() == "vip":
        discount = order_amount * 0.20
    elif customer_type.lower() == "regular" and order_amount >= 500:
        discount = order_amount * 0.10
    elif customer_type.lower() == "new":
        discount = order_amount * 0.15
        if discount > 100:
            discount = 100.0

    if customer_type.lower() == "vip" or (customer_rating >= 4.8 and order_amount > 1000):
        priority = "High Priority"
    elif customer_rating >= 4.0:
        priority = "Medium Priority"
    else:
        priority = "Low Priority"

    if previous_cancellations >= 3 or (payment_method.lower() == "cod" and customer_rating < 3.5):
        cancel_risk = "High"
    elif previous_cancellations >= 1 or payment_method.lower() == "cod":
        cancel_risk = "Medium"
    else:
        cancel_risk = "Low"

    if customer_type.lower() == "vip" and order_amount >= 1500:
        category = "Premium VIP"
    elif cancel_risk == "High":
        category = "High Risk Order"
    elif delivery_distance > 10:
        category = "Long Distance"
    else:
        category = "Normal Order"

    if order_status == "Rejected":
        final_bill = 0.0
    else:
        final_bill = order_amount - discount + total_delivery_fee
        if final_bill < 0:
            final_bill = 0.0

    print("\n--------------------------------")
    print(" ORDER REPORT - DIVYA'S SYSTEM ")
    print("--------------------------------")
    print("Processed By          : Divya")
    print("Order Status          :", order_status)
    print("Restaurant Status     :", restaurant_status)
    print("Manual Review Needed  :", manual_review)
    print("Priority Level        :", priority)
    print("Cancellation Risk     :", cancel_risk)
    print("Category              :", category)
    print("--------------------------------")
    print("Base Amount           : Rs", order_amount)
    print("Discount              : Rs", discount)
    print("Delivery Fee          : Rs", total_delivery_fee)
    print("--------------------------------")
    print("Total Amount Payable  : Rs", final_bill)
    print("--------------------------------")

main()
```
